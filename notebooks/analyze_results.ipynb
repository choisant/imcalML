```python
def get_scores(df):
    print(f'Accuracy: {df["ACC"].mean()*100}% +- {df["ACC"].std()*100}%')
    print(f'Accuracy SPH_9: {df["ACC_SPH_9"].mean()*100}% +- {df["ACC_SPH_9"].std()*100}%')
    print(f'Accuracy BH_n4_M8: {df["ACC_BH_n4_M8"].mean()*100}% +- {df["ACC_BH_n4_M8"].std()*100}%')
    print(f'Accuracy BH_n2_M10: {df["ACC_BH_n2_M10"].mean()*100}% +- {df["ACC_BH_n2_M10"].std()*100}%')
    print(f'Accuracy BH_n4_M10: {df["ACC_BH_n4_M10"].mean()*100}% +- {df["ACC_BH_n4_M10"].std()*100}%')
    print(f'Accuracy BH_n6_M10: {df["ACC_BH_n6_M10"].mean()*100}% +- {df["ACC_BH_n6_M10"].std()*100}%')
    print(f'Accuracy BH_n4_M12: {df["ACC_BH_n4_M12"].mean()*100}% +- {df["ACC_BH_n4_M12"].std()*100}%')
    print(f'Cross entropy: {df["LogLoss"].mean()} +- {df["LogLoss"].std()}')
    print(f'Train time: {df["Train time"].mean()/60} +- {df["Train time"].std()/60} min')
    print(f'Test time: {df["Test time"].mean()} +- {df["Test time"].std()} sec')
```

```python
get_scores(xgboost_hl_df)
```

```python
get_scores(xgboost_ll_df)
```
